```python
import plotly.express as px
```

```python
df = px.data.stocks()
df
```

```python
fig = px.line(df,x='date', y=['GOOG','AAPL'])
fig
```

```python
fig = px.line(df,x='date', y=['GOOG','AAPL'])
fig.update_layout(title ='Stock prices over time',
                  xaxis_title='Date',
                  yaxis_title='Price')
```

```python
fig = px.line(df,x='date', y=['GOOG','AAPL'],
             markers=True,
             )
fig.update_layout(title ='Stock prices over time',
                  xaxis_title='Date',
                  yaxis_title='Price',
                 showlegend=True,
                 plot_bgcolor='#72cad4',
                 hovermode='x unified',
                 template='ggplot2')
```
