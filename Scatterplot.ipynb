```python
import seaborn as sns
```

```python
df = sns.load_dataset('diamonds')
df
```

```python
sns.scatterplot(df, x='carat', y='price')
```

```python
filt = (df['cut'].isin(['Premium', 'Very Good', 'Good'])) & (df['clarity']=='SI1')
df = df.loc[filt]
df
```

```python
df = df.sample(100)
df
```

```python
sns.scatterplot(df, x='carat', y='price')
```

```python
sns.scatterplot(df, x='carat', y='price', hue='cut')
```

```python
sns.scatterplot(df, x='carat', y='price', style='cut')
```

```python
sns.scatterplot(df, x='carat', y='price', size='cut')
```

```python
sns.scatterplot(df, x='carat', y='price', hue='cut', alpha = 0.5)
```
